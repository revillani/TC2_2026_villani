### Conclusión
A lo largo de este trabajo se verificó que una única topología general, un bloque comb en cascada con una red recursiva de segundo orden (biquad), es lo suficientemente flexible como para sintetizar filtros digitales de comportamiento muy distinto con solo activar o anular los coeficientes adecuados. La baraja de posibles filtros es tan diversa que incluye tanto filtros IIR, como es el caso del *DC Blocker*, como FIR, por ejemplo, el *filtro de media móvil* o el *filtro peine*.

En definitiva, el ejercicio muestra cómo el análisis de polos y ceros en el plano Z es la herramienta central para predecir tanto la estabilidad como la forma de la respuesta en frecuencia de cualquier filtro derivado de esta estructura común.

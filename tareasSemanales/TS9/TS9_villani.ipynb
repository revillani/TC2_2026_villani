### Punto 1 (y Bonus 1)

La sintesis de cuadripolos simplemente cargados consiste en tomar a la función transferencia y darle el formato:

### <center>$T(s) = \frac{F.T.}{1+F.E.}$</center>

**Advertencia**: Es fundamental que la transferencia del numerador sea una función **impar**.

Para ello se toma denominador común con la parte par o impar segun convenga:

- ### $impar = \frac{impar}{par} = \frac{par}{impar}$

Después se trabaja la transferencia según la carga para que esta quede en función de una F.T. y una F.E. de parametros $Z$ o $Y$ siguiendo el formato propuesto.

Por último se hacen las remociones de los polos de la F.E. en las ubicaciones de los ceros de la F.T.

Recordar que *una remoción de un polo en una F.E. se traduce en un cero de transmisión*.

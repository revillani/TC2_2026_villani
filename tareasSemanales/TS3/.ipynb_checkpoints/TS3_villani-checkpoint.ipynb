### Punto 1

Por lo visto en la teoría sabemos que:
### <center>$T(s) = \frac{1}{cosh(s)+senh(s)}$</center>

Para obtener el denominador usamos el método de *Stoch* que consiste en trabajar aproximando a la cotangente hiperbólica.

# <center>$T(s) = \frac{cosh(s)}{senh(s)} = \frac{1}{s} + \frac{1}{\frac{3}{s}+\frac{1}{\frac{5}{s}+\frac{1}{\frac{7}{s}+...}}}$</center>

Luego se completa el numerador con el término independiente del denominador para que la ganancia sea de 0 dB.

Para obtener la transferencia de ***mínimo orden*** que cumpla con la condición tenemos que iterar el cálculo de atenuación sobre $\omega_p$ para cada filtro.

Se considera que **la atenuación es la inversa de la transferencia** $\alpha_{máx} = \frac{1}{|T(\omega_p)|}$.

Además $|T(\omega)| = T(s)|_{s=j\omega}$.

### Punto 3

Para evaluar el retardo, primero necesito la ubicación de cada polo.

```python
# Importo las librerías para trabajar de acá en adelante.

# Librerías externas NumPy, SciPy y Matplotlib
from scipy.signal import TransferFunction
from scipy import signal as sig
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp

# Librería de TC2
from pytc2.sistemas_lineales import pzmap, GroupDelay, bodePlot, analyze_sys, tf2sos_analog, pretty_print_lti
from pytc2.general import a_equal_b_latex_s, print_latex
```

```python
order = 3

z,p,k = sig.besselap(order, norm='delay')
num, den = sig.zpk2tf(z,p,k)
T = TransferFunction(num, den)

poles = np.roots(den)
print(poles)
pzmap(T)
```

Polos:
- $s = -1.83890732 \pm 1.75438096j$
- $s = -2.32218535$

El retardo total será la suma de los retardos aportados por cada singularidad.

<center><img src="./src/img6_TS3.png" width="800"></center>

- El Retardo de Grupo $D(2.5)$ es de $0,752$
- El deviamiento porcentrual respecto a $D(0)$ es $\Delta D\% = 24,8\%$.

### Punto 4

Primero uso la ubicación de los polos para formar las trasferencias de cada etapa.

```python
poles1 = np.array([
    -1.83890732 + 1.75438096j,
    -1.83890732 - 1.75438096j
])

poles2 = np.array([
    -2.32218535+0.j
])

zeros = [] 

num1, den1 = sig.zpk2tf(zeros, poles1, 1)
num2, den2 = sig.zpk2tf(zeros, poles2, 1)

# Fijo los numeradores para 0dB
num1 = den1[2]
num2 = den2[1]

# Contruyo las expresiones
s = sp.symbols("s");

num1_expr = num1
den1_expr = s**2 + s*den1[1] + den1[2]
num2_expr = num2
den2_expr = s + den2[1]

T1_expr = num1_expr / den1_expr
T2_expr = num2_expr / den2_expr

print_latex(a_equal_b_latex_s('T1(s)', T1_expr))
print_latex(a_equal_b_latex_s('T2(s)', T2_expr))
```

### Bonus 2

Desnormalizo los componentes para que $D(0) = 200microseg$.

La desnormalización para fijar valor de retardo para $\omega = 0$ consiste en **escalar el eje de frecuencia para reducir la pendiente de la fase (el retardo)**.

Recordar que al obtener la función de aproximación:

#### $T(j\omega) = 1 * e^{-j\frac{\omega}{1/t_o}}$

$\frac{1}{t_o}$ es una norma de frecuencia. Por tanto:

#### $s = \frac{S}{1/D} = \frac{S}{1/200\mu s} = \frac{S}{5000}$

Obtengo la transferencia desnormalizada:

### $T(S) = \frac{6,459}{(\frac{S}{5000})²+3,678*\frac{S}{5000}+6,459} * \frac{2,322}{\frac{S}{5000}+2,322} = \frac{1,61475*10⁸}{S²+18390*S+1,61475*10⁸} * \frac{11610}{S+11610}$

Calculo los componentes para cada etapa.

Hago un escalado de impedancias para que los valores de los capacitores sean $C_{4A} = C_{4B} = 10nF$ y $C_{3A} = 100nF$

$C_{4A} = C_{4B} = 1 = \cancel{\Omega_\omega}*\Omega_z*10nF  \space\space\space\space\space\space \rightarrow \space\space\space\space\space\space \Omega_z = 100M\Omega$
